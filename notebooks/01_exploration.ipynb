```python
import pandas as pd

# Si erreur d'encodage UTF-8, remplace par encoding="latin-1"
df = pd.read_csv("../data/raw/superstore_raw.csv", encoding="latin-1")

print(f"Dimensions : {df.shape[0]} lignes, {df.shape[1]} colonnes")
df.head()
```

```python
# Types de colonnes
df.dtypes
```

```python
# Vérifier si des codes postaux commencent par 0 (auquel cas int64 les tronque)
print(df["Postal Code"].astype(str).str.len().value_counts())
```

```python
# Regarder quelques exemples de codes à 4 chiffres, avec leur État
codes_4 = df[df["Postal Code"].astype(str).str.len() == 4]
codes_4[["State", "City", "Postal Code"]].drop_duplicates().head(10)
```

```python
# Valeurs manquantes par colonne
df.isnull().sum().sort_values(ascending=False)
```

```python
# Doublons
print(f"Lignes dupliquées : {df.duplicated().sum()}")
```

```python
# Vérification de cohérence : un Customer ID a-t-il toujours le même Customer Name ?
incoherences = df.groupby("Customer ID")["Customer Name"].nunique()
print(f"Customer ID avec plusieurs noms différents : {(incoherences > 1).sum()}")
```
